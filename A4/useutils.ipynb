```python
# First, ensure you have saved the previous code block as 'geoutils.py'
# in the same directory as this Jupyter Notebook, or in a directory
# included in your Python path.

# Import the functions from your geoutils.py file
from geoutils import calculate_distance, is_within_radius, get_bearing
```

```python

# --- Example Usage of calculate_distance ---
# Let's calculate the distance between New York City and London
nyc_lat, nyc_lon = 40.7128, -74.0060
london_lat, london_lon = 51.5074, -0.1278

distance = calculate_distance(nyc_lat, nyc_lon, london_lat, london_lon)
print(f"Distance between NYC and London: {distance:.2f} km")
print("-" * 30)
```

```python

# --- Example Usage of is_within_radius ---
# Check if a point (e.g., a restaurant) is within 5 km of a user's location
user_lat, user_lon = 48.8566, 2.3522  # Paris
restaurant_lat, restaurant_lon = 48.8584, 2.2945 # Eiffel Tower (close to Paris)
search_radius_km = 5

within_radius = is_within_radius(restaurant_lat, restaurant_lon, user_lat, user_lon, search_radius_km)
print(f"Is the Eiffel Tower within {search_radius_km} km of central Paris? {within_radius}")

# Let's try a point further away, e.g., Rome
rome_lat, rome_lon = 41.9028, 12.4964
within_radius_rome = is_within_radius(rome_lat, rome_lon, user_lat, user_lon, search_radius_km)
print(f"Is Rome within {search_radius_km} km of central Paris? {within_radius_rome}")
print("-" * 30)

```

```python
# --- Example Usage of get_bearing ---
# Calculate the bearing from Lisbon to Madrid
lisbon_lat, lisbon_lon = 38.7223, -9.1393
madrid_lat, madrid_lon = 40.4168, -3.7038

bearing = get_bearing(lisbon_lat, lisbon_lon, madrid_lat, madrid_lon)
print(f"Initial bearing from Lisbon to Madrid: {bearing:.2f} degrees")
print("-" * 30)

# Calculate the bearing from North Pole to South Pole (roughly South)
north_pole_lat, north_pole_lon = 90.0, 0.0
south_pole_lat, south_pole_lon = -90.0, 0.0

bearing_poles = get_bearing(north_pole_lat, north_pole_lon, south_pole_lat, south_pole_lon)
print(f"Initial bearing from North Pole to South Pole: {bearing_poles:.2f} degrees (should be ~180)")
```
